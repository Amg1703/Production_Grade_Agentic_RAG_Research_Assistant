- WHERE .MAJOR WILL BE THE FIRST DIGIT .MINOR WILL BE THE SECOND DECIMAL PLACE AND MICRO WILL BE THE 3RD DECIMAL PLACE OF THE PYTHON VERSION

### WE ARE CHECKING THE SYS EXECUTABLE 
- TO CHECK THE WHERE THE PYTHON.EXE EXECUTABLE EXISTS SINCE WE WANT THIS EXECUTABLE TO RUN ON UV/SCRIPTS BUT IF WE HAD OTHER PYTHON VERSION INSTALLED IN OUR ENVIRONMENT THERE MIGHT BE POSSIBLE CASES WHERE THE PYTHON EXECUTABLE RUNNING BELONGS TO IT 

```python
import sys

python_version=sys.version_info
print(f"Python version:{python_version.major}.{python_version.minor}.{python_version.micro}")
print(f"Environment:{sys.executable}")

if python_version>=(3,12):
    print("Python version compatable")
else:
    print("Python version not compatable, since python version < 3.12")
```

```python
from pathlib import Path

current_dir=Path.cwd()
if current_dir.name=="week1" and current_dir.parent.name=="notebook":
    project_dir=current_dir.parent.parent
elif (current_dir/"compose.yml").exists():
    project_dir=current_dir
else:
    project_dir=None

if project_dir and (project_dir/"compose.yml").exists():
    print(f"Project root:{project_dir}")
else:
    print("Project root not found or is not at the right directory")
    exit()
```

```python
import subprocess

try:
    result=subprocess.run(["docker", "--version"], capture_output=True, text=True, timeout = 5)
    if result.return_code==0:
        print(f"Docker working: {result.stdout}")
    else:
        print(f"Docker not working: {result.stderr}")
        exit()
except:
    print("Docker not found")
    exit()
```

```python
try:
    result=subprocess.run(["docker", "compose", "version"], capture_output=True, text=True, timeout=5)
    if result.return_code==0:
        print(f"Docker compose working: {result.stdout.split()[3]}")
    else:
        print(f"Docker compose is not working {result.stderr}")
        exit()
except:
    print("Docker compose not found")
    exit()
```

```python
try:
    result=subprocess.run(["uv","--version"], capture_output=True, text=True, timeout=5)
    if result.return_code==0:
        print(f"UV working: {result.stdout.strip()}")
    else:
        print(f"UV is not working {result.stderr}")
        exit()
except:
    print("UV is not found")
    exit()
    
```

```python
try:
    result=subprocess.run(["docker", "info"], capture_output=True, text=True, timeout=5)
    if result.return_code==0:
        print(f"Docker daemon is working: {result.stdout}")
    else:
        print(f"Docker daemon is not working: {result.stderr}")
        exit()

except:
    print(f"Docker daemon is not accessible")
    exit()
```
